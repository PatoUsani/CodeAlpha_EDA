```python
# 1. Imports
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# Set style
sns.set_theme(style="whitegrid")

# 2. Load Dataset
df = pd.read_csv("sales_data.csv")

# ---------------------------------------------------------
# STEP 1: Data Inspection
# ---------------------------------------------------------
print("Head of Data:")
display(df.head())

print("\nData Info:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

# ---------------------------------------------------------
# STEP 2: Data Cleaning
# ---------------------------------------------------------
# Drop duplicates if any
df = df.drop_duplicates()

# Handle missing values (example: fill numerical missing with median)
num_cols = df.select_dtypes(include=[np.number]).columns
for col in num_cols:
  df[col] = df[col].fillna(df[col].median())

# ---------------------------------------------------------
# STEP 3: Descriptive Statistics
# ---------------------------------------------------------
display(df.describe())

# ---------------------------------------------------------
# STEP 4: Univariate Analysis (Distribution)
# ---------------------------------------------------------
plt.figure(figsize=(8, 4))
sns.histplot(df[num_cols[0]], kde=True, color="indigo")
plt.title(f"Distribution of {num_cols[0]}")
plt.savefig("univariate_plot.png", bbox_inches="tight")
plt.show()

# ---------------------------------------------------------
# STEP 5: Bivariate / Correlation Analysis
# ---------------------------------------------------------
plt.figure(figsize=(10, 6))
sns.heatmap(df[num_cols].corr(), annot=True, cmap="mako", fmt=".2f")
plt.title("Correlation Matrix")
plt.savefig("correlation_matrix.png", bbox_inches="tight")
plt.show()
```
